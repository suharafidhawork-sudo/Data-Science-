```python
with open("EX1.txt","r+") as f:#----->When opened, the pointer starts at position 0 (the start of the file)
    f.write("hi barbie")#----->overwrites the first 9 characters of the original string ("HELLO WOR" becomes "hi barbie")
    cont=f.read()#----------->After writing 9 bytes, the file cursor sits immediately after "barbie". Calling f.read() reads everything remaining from that point to the end of the file.
    print(cont)
```
