```python
import matplotlib.pyplot as plt
```

```python
#TYPES OF PLOTS
#Line plot
x=[1,2,3,4]
y=[5,7,9,11]
plt.plot(x,y)#connects both points x and y
plt.show()# to display the plot
```

```python
#adding more parameters
plt.figure(figsize=(2,2))#size of the plot graph
plt.plot(x,y,color='g',linestyle='-',linewidth=3,marker='*',markersize=8)
plt.title("LINE PLOT",color='r')
plt.xlabel('x axis',color='r')
plt.ylabel('y axis',color='r')
plt.show()
```

```python
#Area Plot
plt.figure(figsize=(2,2))
plt.fill_between(x,y,color='g',alpha=0.3)
plt.title('AREA PLOT')
plt.xlabel('x axis',color='r')
plt.ylabel('y axis',color='r')
plt.show()



```

```python
month=['jan','feb','march','april','may','june']
SALE_A=[22200,3450000,45000,60000,7000,600000]
SALE_B=[3400000,460000,500,50000,60000,4000]
plt.figure(figsize=(5,5))
plt.fill_between(month,SALE_A,SALE_B,alpha=0.5,color='c')
plt.plot(month,SALE_A)
plt.plot(month,SALE_B)
```

```python
#SCATTER PLOT
# X-axis: Hours spent studying
# Y-axis: Resulting exam scores (%)
x = [1.5, 2.0, 3.0, 4.0, 4.5, 5.5, 6.0, 7.0, 8.0, 9.0]
y = [40, 48, 55, 62, 65, 75, 78, 85, 92, 96]
plt.figure(figsize=(4,4))
plt.scatter(x,y,s=100,c=y,marker="o",cmap='magma')
plt.colorbar(label='color intensity')
plt.title("SCATTER PLOT",color='r')
plt.xlabel("hours spent learning",color='m')
plt.ylabel("results in exam scores",color='y')
```

```python
# Dataset 1: Morning Batch (Study Hours vs Exam Scores)
x_morning = [2.0, 3.5, 4.0, 5.5, 6.5, 7.5, 8.0]
y_morning = [50, 60, 68, 75, 82, 88, 95]

# Dataset 2: Evening Batch (Study Hours vs Exam Scores)
x_evening = [1.5, 2.5, 3.8, 4.5, 6.0, 7.0, 8.5]
y_evening = [45, 52, 63, 70, 79, 85, 91]
plt.figure(figsize=(5,5))
plt.scatter(x_morning, y_morning, color='red', label='Morning Batch')
plt.scatter(x_evening, y_evening, color='c', label='Evening Batch')
plt.title("SCATTER PLOT",color='m')
plt.xlabel("study_hours",color='r')
plt.ylabel("exam_score",color='g')
plt.legend()
plt.show()
```

```python
#plot by category
x=[11,12,13,14,15]
y=[20,30,40,50,60]
cat=['a','b','a','b','a']
colors=['red' if c =='a' else 'blue' for c in cat]
plt.scatter(x,y,color=colors)
plt.grid(True)
plt.legend()
```
