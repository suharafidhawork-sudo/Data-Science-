```python

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
d=sns.load_dataset('tips')
#BARPLOT
sns.barplot(x='day',y='tip',data=d,ci=95,hue='sex',palette='rainbow')
plt.title("BARPLOT")
plt.show()
```

```python
#COUNT PLOT
#used to count the number of observations
#used only for categorical values
plt.figure(figsize=(10, 6))
sns.countplot(data=d, x='day', hue='sex')  # Using 'day' as an example of a categorical column
plt.show()
```

```python
#BOX PLOT
sns.boxenplot(x='day',y='total_bill',hue='sex',data=d)
plt.title("BOX PLOT")
```

```python
#just one numberical variabledistribution
sns.boxenplot(x='total_bill',data=d,palette='Set2')
plt.show()
```

```python
#VIOLIN PLOT

sns.violinplot(x='day',y='total_bill',hue='sex',data=d,split=True)
plt.title("VIOLIN PLOT")
```

```python
#STRIP PLOT
sns.stripplot(x='day',y='tip',data=d,hue='sex',jitter=True,dodge=True,edgecolor='m',size=7)
plt.title("STRIP PLOT")
```

```python
#CAT PLOT
sns.catplot(x='day',y='tip',col='time',data=d,kind='strip',hue='sex',palette='Set2')
plt.title("CAT PLOT")
```

```python
sns.catplot(x='day',y='tip',col='time',data=d,kind='point',hue='sex',palette='Set2')
```

```python
sns.catplot(x='day',y='tip',col='time',data=d,kind='box',hue='sex',palette='Set2')
```

```python
#matrix plot
d.corr(numeric_only=True)#only shows the numeric variable
```
