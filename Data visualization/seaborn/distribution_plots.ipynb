## SEABORN

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()#too get the names of the in built data set
```

```python
data=sns.load_dataset('diamonds')#---> to load adata set into a variable
data
```

```python
#DISTRIBUTION PLOTS
d=sns.load_dataset('tips')
sns.histplot(data=d,x=d['total_bill'],bins=20,kde=True,hue=d['sex'],palette='rainbow')#-- it vizualize the distiribution of total bill categorized on sex 
#kde--->kernel density line used to represent the smoothness of the distribution
```

```python
#KDE PLOT
sns.kdeplot(data=d,x=d['total_bill'],fill=True,hue=d['sex'],bw_adjust=1,multiple='layer')
```

```python
#DISPLOT
sns.displot(data=d, x='tip', col="sex", kde=True)
plt.show()
```

```python
sns.displot(data=d, x='tip', col="sex",kind='ecdf')
plt.show()
```

```python
#JOIN PLOT
sns.jointplot(x='total_bill',y='tip',data=d,kind='scatter',color='r')
```

```python
#RUGPLOT
sns.rugplot(data=d['tip'],height=0.5,color='g')
plt.title("RUGPLOT")
plt.ylabel("frequency")
```
