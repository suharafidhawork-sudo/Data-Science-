```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
d=sns.load_dataset('tips')
```

```python
#HEATMAP
corr=d.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap='magma',linewidths=0.5)
```

```python
#PAIR PLOT
sns.pairplot(d,hue='sex',diag_kind='kde',kind='scatter')
```
