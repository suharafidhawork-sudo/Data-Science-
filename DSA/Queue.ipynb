```python
class Queue:
    def __init__(self):
        self.queue=[]
    def enqueue(self,d):
        self.d=d
        self.queue.append(self.d)  
    def dequeue(self):
        if len(self.queue)==0:
            print("stack is empty")
        else:
            return self.queue.pop(0)
    def display(self):
        print(self.queue)  
         
               
```

```python
q=Queue()
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)
q.display()
q.dequeue()
q.display()
```
