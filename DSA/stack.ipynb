```python
class Stack:
    def __init__(self):
        self.stack=[]#---------->creating an empty list
    def push(self,d):
        self.d=d
        self.stack.append(self.d)  
    def pop(self):
        if len(self.stack)==0:
            print("stack is empty")
        else:
            return self.stack.pop()
    def display(self):
        print(self.stack)   
```

```python
s=Stack()
s.push(23)#--------->adding an element to the stack
s.push(4)
s.push(20)
s.display()
s.pop()#---------->deleting an element in the stack
s.display()
```
