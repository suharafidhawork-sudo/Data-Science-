#### ELEMENTS=10,7,15,20,5,2

### TREE 

```python
#BINARY TREE

class Node:
    def __init__(self,data):
        self.data=data
        self.right=None
        self.left=None
    
        

```

```python
root = Node(10)
```

```python
root.left=Node(7)
root.left.left=Node(15)
root.left.left=Node(2)
root.right=Node(15)
root.right.right=Node(20)
```

```python
def Inorder(node):
    if node:
        Inorder(node.left)
        print(node.data,end=" ")
        Inorder(node.right)
```

```python
def preorder(node):
    if node:
        print(node.data,end=" ")
        preorder(node.left)
        preorder(node.right)
```

```python
def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data,end=" ")
```

```python
Inorder(root)
```

```python
preorder(root)
```

```python
postorder(root)
```
