```python
def bubble_sort(l):
    n=len(l)
    for i in range(n):
        for j in range((n-1)):
            
            if l[j]>l[j+1]:
                l[j],l[j+1]=l[j+1],l[j] 
    return l                     
```

```python
num=[1,7,3,4,6]
bubble_sort(num)
```

```python
def selection_sort(a):
    n=len(a)
    for i in range(n):
        min=i
        for j in range(i+1,n):
            if a[j]<a[min]:
                min=j
                
        a[i],a[min]=a[min],a[i]     
    return a       
            
            
```

```python
num=[3,6,2,1,5]
selection_sort(num)
```

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]        # element we're placing correctly
        j = i - 1
        while j >= 0 and arr[j] > key:  # shift larger elements right
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key      # place key in correct spot
    return arr
```

```python
def merge_sort(arr):
    if len(arr) <= 1:       # base case — single element is sorted
        return arr
    
    mid = len(arr) // 2
    left  = merge_sort(arr[:mid])   # sort left half recursively
    right = merge_sort(arr[mid:])   # sort right half recursively
    
    return merge(left, right)       # merge both sorted halves

def merge(left, right):
    result = []
    i = j = 0
    # compare elements from both halves one by one
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    # add remaining elements
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

```python
def quick_sort(arr):
    if len(arr) <= 1:       # base case
        return arr
    
    pivot = arr[len(arr) // 2]  # pick middle element as pivot
    
    left   = [x for x in arr if x < pivot]   # smaller than pivot
    middle = [x for x in arr if x == pivot]  # equal to pivot
    right  = [x for x in arr if x > pivot]   # larger than pivot
    
    # recursively sort left and right, join all three
    return quick_sort(left) + middle + quick_sort(right)
```

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:    # found it
            return i            # return its position
    return -1                   # not found
```

```python
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1
    
    while left <= right:
        mid = (left + right) // 2   # find middle index
        
        if arr[mid] == target:      # found
            return mid
        elif arr[mid] < target:     # target is in right half
            left = mid + 1
        else:                       # target is in left half
            right = mid - 1
    
    return -1   # not found
```
