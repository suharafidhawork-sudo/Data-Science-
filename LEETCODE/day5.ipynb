```python
def product(nums):
    answers=list()
    
    for i in range(len(nums)):
        p=1
        for j in range(0,len(nums)):
            
            if nums[i]!=nums[j]:
                p=p*nums[j]
            if nums[j]==0:
                p=0    
        answers.append(p)    
    return print(answers)        
        
```

```python
n=[0,0]
product(n)
```
