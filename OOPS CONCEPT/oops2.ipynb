### HEIRACHICAL INHERITENCE


```python
class Doctor:
    def __init__(self,name,specialization):
        self.name=name
        self.specialization=specialization
    def consult(self):
        print("Doctor: ",self.name,"\nSpecialization: ",self.specialization)  
        print(self.name,"is consulting ")
class Surgeon(Doctor):
    def __init__(self, name, specialization):
        super().__init__(name, specialization)
    def operate(self):
            print(self.name,"is performing surgery")   
    
class Pediatrician(Doctor):
    
    def __init__(self, name, specialization):
        super().__init__(name, specialization)  
    def treat(self):
            print(self.name,"is treating a child patient") 


                         
       
        
        
```

```python
d1=Doctor("fidha","pediatrician")
d1.consult()
```

```python
d2=Surgeon("riri","surgeon")
d2.operate()
```

```python
d3=Pediatrician("riri","pediatrician")
d3.treat()
```
