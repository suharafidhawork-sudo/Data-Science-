```python
class Test:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):#------>method overloading my
        return a+b+c
```

```python
v1=Test()
v1.add(2,3)#----->method overloading python dosen't support method overloading
```

```python
class Vehicle:
    def start(self):
        print("vehicle is starting")
class Bike(Vehicle):
    def start(self):
        print("bike is starting")   
class Car(Vehicle):
    def star(self):
        print("car is starting")     
        
```

```python
l=[Vehicle(),Bike(),Car()]#---->evry class has same method name but it only works when that object that specific class is being called , therefor pyhton supports overrriding
for i in l:
    i.start()
```

```python
class BankAccount:
    def __init__(self,name ,accountno,balance):
        self.name=name
        self._accountno=accountno#--------->protected members "_"--> adding before the variable name
        self._balance=balance
    def details(self):
        print("ACCOUNT NO: ",self._accountno)
        print("BALANCE:",self._balance)
        
        
```

```python
f1=BankAccount("fidha","3456","34000")
f1.details()
f1._accountno=567899
f1.details()
```
