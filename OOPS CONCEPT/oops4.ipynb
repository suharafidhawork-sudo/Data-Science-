```python
from abc import  ABC, abstractmethod
```

```python
#Abstraction\
    
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class UPI(Payment):
    def _init_(self, pin , balance):
        self.__pin = pin
        self.__balance = balance

    def pay(self, amount):

        if not self.__check_pin():
            print("Payment Failed")
            return
        
        if not self.__check_balance(amount):
            print("Payement Failed")
            return
        
        self.__connect_bank()
        self.__payment_status(amount)

    def __check_pin(self):
        entered_pin = int(input("Enter the pin :"))

        print("Checking UPI Pin.....")
        if entered_pin == self.__pin:
            print("Pin Verified")
            return True

        print("Incorrect Pin")
        return False
    
    def __check_balance(self, amount):
        print("Checking Bank Balance.....")
        if self.__balance >= amount:
            return True

        print("Insufficient Balance")
        return False
    
    def __connect_bank(self):
        print("Connecting to Bank.....")

    def __payment_status(self, amount):
        self.__balance -= amount
        print(f"${amount} paid successfully.....")
        print(f"Remaining Balace : ${self.__balance}")
```
