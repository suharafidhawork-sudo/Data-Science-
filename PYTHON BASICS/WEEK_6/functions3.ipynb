```python
#q1
#MAX OF THREE NUMBERS

def max_num(a,b,c):
    print("MAX OF ",a,",",b,",",c,":",max(a,b,c))
num1=int(input("enter th first number: "))
num2=int(input("enter the second number: "))
num3=int(input("enter the third number: "))  
max_num(num1,num2,num3)      
```

```python
#Q2
#FACTORIAL OF NUMBERS

def fact(n):
    fact=1
    i=1
    while i<=n:
        fact*=i
        i+=1
    print("FACTORIAL OF ",n,":",fact)   
num=int(input("enter the number:"))    
fact(num)
```

```python
#Q3
#prime numbers
def is_prime(n):
    i=1
    count=0
    while i<=n:
        if n%i==0:
            count+=1
        i+=1
    if count==2:
        return True
    else:
        return False 
            
num=int(input("enter the number:"))
is_prime(num)            
            
```

```python
#Q3
#FIBONACCI SERIES

def fibonacci(n):
    a=0
    b=1
    while a<=num:
        print(a, end=" ")
        a,b=b,a+b

num=int(input("enter the number :"))
fibonacci(num)              

        
```

```python
#vowels

def vowels(s):
    count=0
    for i in s:
        if i in 'AaEeIiOoUu':
            count+=1
    print("TOTAL NUMBER OF VOWELS IN STRING :",count)        
a=input("enter the string: ")
vowels(a)            
```

```python
#COUNT OF CHARACTER
def count_string(s):
    b=dict()
    for i in set(s):
        print(i,"=",s.count(i))
        b[i]=s.count(i)
    print(b)    
           
s=input("enter the string")  
count_string(s)     
```

```python
def count_word(s):
    count=0
    b=s.split()
    for i in b:
        count+=1
    print(count)    
s=input("enter the string:")
count_word(s)        
```

```python
#MIDDLE ELEMENT

def mid(s):
    l=len(s)
    mid=l//2
    print(s[mid])

s=[1,2,3,4,5]
mid(s)
 
```

```python
def remove_negative(s):
    for i in s:
        if i<0:
            s.remove(i)
    print(s) 
s=[1,2,-4,5,7]
remove_negative(s)  



         
```

```python
#OCCURENCES
def occurence(s,a,r):
    b=list(s)
    
    for i in range(len(b)):
        if b[i]==a:
            b[i]=r
    print("".join(b)) #JOIN THE STRING        
            
    
      
            
s=input("enter the string :")
a=input("enter the character that is to bee replaced: ")
r=input("enter the new character")   
occurence(s,a,r)            
```

```python
#SUM OF N NATURAL PRIME NUMBERS



def is_prime(n):
    i=1
    count=0
    while i<=n:
        if n%i==0:
            count+=1
        i+=1
    if count==2:
        return True
    else:
        return False 

def sum_prime(n):
    sum=0
    for i in range(1,n+1):
        if is_prime(i)==True:
            sum+=i
    print("sum of prime numbers up to the range of",n,":",sum)
n=int(input("enter the range :"))
sum_prime(n)         
            
            
```
