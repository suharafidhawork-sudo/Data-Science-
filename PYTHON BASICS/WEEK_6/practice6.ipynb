```python
def rev(s):
    if len(s)<=1:
        return s
    return s[-1]+rev(s[:-1])   
a="fidha"
# rev(a)
print(a[:-1])
print(a[-1])
 
```

```python
def palindrome(s):
    if len(s)==1:
        return True
    if s[0]!=s[-1]:
        return False
    return s[-1]+palindrome([1,-1])
a="fidha"
palindrome(a)
```

```python
def power(n,e):
    if e==0:
        return 1
    return n*power(n,e-1)
power(2,3)
```

```python
def amstrong(n,og=n,t=0):

    if  n==0:
        return t==og
    
    
    d=len(str(og))
    r=n%10
    f=r**d
    return amstrong (n//10, og , t+f)

print(amstrong(153))
print(amstrong(101))

    
    
```

```python
def sum_list(l,sum=0,i=0):
    if i==len(l):
        return sum
    return sum_list(l,sum=sum+l[i],i=i+1)
a=[1,2,3]
sum_list(a)
```

```python
def max(l,max=0,i=0):
    if l[i]==max and i==len(l):
        return max
    max=l[i]
    if l[i+1]>max:
        m=l[i+1]
    return max(l,max,i=i+1)  

a=[2,3,1]  
max(a)
    
```

```python
def factors(n,l=list(),i=1):
    if i==n+1:
        return l
    if n%i==0:
        l.append(i)
    return factors(n,l,i=i+1)  

factors(12)
        
    
```

```python
def flatten(l,li=list(),i=0):

    if i==len(l):
        return li
    for j in l[i]:
        li.append(j)  
    return flatten(l,li,i=i+1)
b=[[1,3],[5,6],[3,8]]    
flatten(b)
```
