```python
#FACTORIAL WITH RETURN STATEMENT

def fact(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact #return statement used to return the value

num=int(input("enter the number:"))
print(fact(num))
```

```python
#employee details using functions

def employee_details(n,id,sal,role):
    d={"name":n,"id":id,"sal":sal,"role":role}
    return d#returns the value as a dictionary

def delete_details(b,n):
    b.remove(n)
    
    
    

b = []
for i in range(3):#STORE DETAILS OF 3 EMPLOYEEE
    name=input("enter the name :")
    id=int(input("enter the id:")) 
    sal=int(input("enter the salary:"))
    role=input("enter the role:") 
    e=employee_details(name,id,sal,role)#storing the value to an variable
    b.append(e)#appending dictionary into a list 
    

for i in b:
    print("name:",i["name"])  
    print("ID",i["id"])
    print("SALARY:",i["sal"])
    print("ROLE: ",i["role"])
    print("*"*40) 
    
```

```python
n=input("enter the name to be deleted:")
for i in b:
    if i['name']==n:#accessing the dictionary element in the list so using "name " in string since the key label is "name "
        delete_details(b,i)#DETELE THE DICTIONARY ELEMENT IN THE LIST B WHERE VALUE OF NAME = N(input from the user)
         
for i in b:
    print("name:",i["name"])  
    print("ID",i["id"])
    print("SALARY:",i["sal"])
    print("ROLE: ",i["role"])
    print("*"*40) 
         
    
```
