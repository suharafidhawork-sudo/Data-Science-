```python
def second_largest_element(l):
    max=highest_element(l)#8
    c=l.index(max)#2
    max2=l[0]
    for i in range(len(l)):#2
        if l[i]<max and i!=c:
            if l[i]>max2:
                max2=l[i]
    return max2



r=int(input("enter the range : "))    
b=create_list(r)
print(b)
second_largest_element(b)
    
    
        
```

```python
def second_least_element(l):
    min=least_element(l)
    c=l.index(min)
    min2=l[0]
    for i in range(len(l)):#2
        if l[i]>min and i!=c:
            if l[i]<min2:
                min2=l[i]
    return min2

r=int(input("enter the range: "))
d=create_list(r)
print(d)
second_least_element(d)    
```

```python
def list_sort(l):
    a=len(l)#4
    for i in range(a):#1 i=0
        for j in range(0,a-i-1):#j(0,3)
            if l[j]>l[j+1]:#l[0]>l[1]
                l[j],l[j+1]=l[j+1],l[j]
    return l

n=[4,7,2,3,1]
print(list_sort(n))            
                
        
```

```python
#function that takes a list of numbers and returns only the even numbers in a new list.
def even_list(l):
    even=list()
    for i in range(len(l)):
         n=l[i]
         if n%2==0:
            even.append(n)
    return even
 
a=[3,9,2,5,7,4,9,12]      
print("EVEN LIST ",even_list(a))
 
            
```

```python
#OCCURENCE OF A NUMBER
def occurence_num(l):
    a = dict()
    for i in l:
        if i in a:
            a[i] += 1
        else:
            a[i] = 1
    return a

a = [1,5,1,3,5,2,8,2,3]
print(occurence_num(a))

              
                
            
```

```python
#function that takes a nested list and returns a flat list.
def flat_list(l):#convert nested list ---> normal list
    a=list()
    for i in l:
        for j in i:
            a.append(j)
    return a   

a=[[2,3,1],[9,5,6],[7,9]]
flat_list(a)
     
```
