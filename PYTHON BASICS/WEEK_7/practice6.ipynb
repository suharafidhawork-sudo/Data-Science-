```python
def rev(s):
    if len(s)<=1:
        return s
    return s[-1]+rev(s[:-1])   
a="fidha"
# rev(a)
print(a[:-1])
print(a[-1])
 
```

```python
def sum_num(n,sum=0):
    if n==0:
        return sum
    return sum_num(n-1,sum=sum+n)
        
    
sum_num(10)
    
     
```

```python
def power(n,e):
    if e==0:
        return 1
    return n*power(n,e-1)
power(2,3)
```
