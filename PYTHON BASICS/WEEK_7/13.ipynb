```python
number=int(input("enter the number of students"))
n=int(input("enter the number of subjects"))
students=dict()
for i in range(number):
    name = input("enter the name: ")
    m = list()  
    
    for j in range(n):
        a = int(input("enter the marks"))
        m.append(a)
    students[name] = m
print(students) 

 
def total_marks(s):
    total=list()
    for i in s:
        total.append(sum(s[i])) 
    return total 

 
 
total=total_marks(students) 

   
def percentage(t):
    p=list()
    for i in t:
        percentage = (i / (n * 100)) * 100
        p.append(percentage)
    return p

per=percentage(total)


def grade(p) :
    g=list()
    for i in p:
        if i>=90:
            g.append("A")
        elif i<90 and i>=80:
            g.append("B")   
        elif i<80 and i>75:
            g.append("C")   
        else:
            g.append("F") 
    return g  

gr=grade(per)


def status(g):
    s=list()
    for i in g:
        if i=="A"or i=="B" or i=="C" :
            s.append("passed")
        else:
            s.append("failed")    
    return(s)     
stat=status(gr)  



def build_report(students, total, per, gr, stat):
    names = list(students.keys())
    report = dict()
    for i in range(len(names)):
        report[names[i]] = {
            "marks": students[names[i]],
            "total": total[i],
            "percentage": round(per[i], 2),
            "grade": gr[i],
            "status": stat[i]
        }
    return report

report = build_report(students, total, per, gr, stat)
print(report)


def report_card(report):
    print("-"*50)
    for name, i in report.items():
        print("Student Name  : ", name)
        print("Marks         : ", i["marks"])
        print("Total Marks   :", i["total"])
        print("Percentage    : ", i["percentage"])
        print("Grade         : ", i["grade"])
        print("Status        : ", i["status"].upper())
        print("-"*50)
report_card(report)
```
