### DEFAULT ARGUMENTS

```python
def dis_name(n="user"):
    print("my name is: ",n)

dis_name()#prints default value since no parameter is specified by the user
dis_name("fidha")   
```

```python
def exp(n,e=3):#DEFAULT EXPONENT IS SET TO 3
    print("cube of ",n,":",n**e)
num=int(input("enter the number:"))

exp(num,2) 




 

  
```
