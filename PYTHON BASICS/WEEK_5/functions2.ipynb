```python
#Q1
#FACTORIAL
def fact(n):
    fact=1
    i=1
    while i<=n:
        fact=fact*i
        i+=1
    print("factorial of ",n,":",fact)  

num=int(input("enter the number:"))
fact(num)      
        
```

```python
#Q2
#PRIME NUMBERS

def is_prime(n):
    i=1
    count=0
    while i<=n:
        if n%i==0:
            count+=1
        i+=1 
    if count==2:
        print("IS A PRIME NUMBER")   
    else:
        print("NOT A PRIME NUMBER")  

num=int(input("enter the number: "))
is_prime(num)              
```

```python
#Q4
#PALINDROME

def is_palindrome(a):
    rev=a[::-1]
    if rev==a:
        print("is a plaindrome")
    else:
        print("not a palindrome") 
s=input("enter a string:")
is_palindrome(s)           
```
