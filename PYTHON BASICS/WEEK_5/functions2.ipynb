```python
#Q1
#FACTORIAL
def fact(n):
    fact=1
    i=1
    while i<=n:
        fact=fact*i
        i+=1
    print("factorial of ",n,":",fact)  

num=int(input("enter the number:"))
fact(num)      
        
```

```python
#Q2
#PRIME NUMBERS

def is_prime(n):
    i=1
    count=0
    while i<=n:
        if n%i==0:
            count+=1
        i+=1 
    if count==2:
        print("IS A PRIME NUMBER")   
    else:
        print("NOT A PRIME NUMBER")  

num=int(input("enter the number: "))
is_prime(num)              
```

```python
#Q4
#PALINDROME

def is_palindrome(a):
    rev=a[::-1]
    if rev==a:
        print("is a plaindrome")
    else:
        print("not a palindrome") 
s=input("enter a string:")
is_palindrome(s)           
```

```python
#Q8
#function that calculates simple interest given principal, rate, time

def simple_interest(p,r,t):
    si=p*r*t/100
    print("SIMPLE INTEREST = ",si)

p=int(input("enter the principal ammount: "))  
r=int(input("enter the rate : ")) 
t=int(input("enter the time: "))
simple_interest(p,r,t) 
```

```python
#Q9
#function that returns sum of digits of a number

def sum_digits(n):
    sum=0
    while n>0:
        r=n%10
        sum+=r
        n=n//10
    print("sum of digits:",sum)    
    

n=int(input("enter the number: "))
sum_digits(n)    
```

```python
#Q12
#function that returns the sum of first n natural numbers
def sum_natural_num(r):
    sum=0
    for i in range(r+1):
        sum+=i
    print("SUM OF FIRST",r,"NATURAL NUMBERS: ",sum)    
r=int(input("enter the range: "))
sum_natural_num(r)    
```
