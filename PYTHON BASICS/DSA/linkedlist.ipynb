```python
class Node:
    def __init__(self,d):
        self.d=d
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
    def insert(self,data):
        new_node=Node(data)
        if self.head is None:
            self.head=new_node
            new_node.next=self.head
            return
        current_node=self.head
        while current_node.next!=self.head:
            current_node=current_node.next
        current_node.next=self.head 
    def display(self):
        current=self.head
        while current:
            print(current.d,end='->')
            current=current.next
        print("None")         
```
