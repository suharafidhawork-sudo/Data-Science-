```python
import numpy as np
scores = np.array([
    [85, 90, 78, 92, 88],   # Alice
    [45, 55, 60, 40, 50],   # Bob
    [90, 95, 88, 97, 93],   # Charlie
    [60, 58, 70, 65, 62],   # Diana
    [78, 82, 75, 80, 79],   # Eve
    [30, 40, 45, 35, 38],   # Frank
    [95, 92, 90, 98, 96],   # Grace
    [55, 60, 58, 52, 57],   # Henry
    [88, 85, 82, 90, 87],   # Iris
    [70, 75, 68, 72, 71],   # Jack
])
students = np.array(['Alice','Bob','Charlie','Diana','Eve',
                     'Frank','Grace','Henry','Iris','Jack'])
subjects = np.array(['Maths','Science','English','CS','Data Science'])

total=np.sum(scores,axis=1)  
print(total)
avg=np.average(scores,axis=1)
print(avg)
avg_sub=np.average(scores,axis=0)
print(avg_sub)
highest_mark=np.max(total)
print(highest_mark)  
rank=np.sort(total)[::-1]      
print(rank)      
        
        
    
    







```

```python
print("*" * 50)
print("     STUDENT PERFORMANCE ANALYSER")
print("*" * 50)
print("-" * 50)
print("TOTAL MARKS OF STUDENTS")
print("-" * 50)
for i in range(len(students)):
    print(students[i],":",total[i])
print("-" * 50)
print("AVERAGE MARKS OF STUDENTS")
print("-" * 50)
for i in range(len(students)):
    print(students[i],":",avg[i])
print("-" * 50)
print("AVERAGE OF EACH SUBJECTS")
for i in range(len(subjects)):
    print(subjects[i],":",avg_sub[i])
print("-" * 50) 
print("HIGHEST MARK :",highest_mark) 
print("-" * 50)
print("LOWEST MARK: ",np.min(total))  
print("-" * 50)
print("STUDENT WHO SCORED HIGHEST MARK ")
print(students[np.argmax(total)],":",highest_mark)
print("-" * 50)
print("STUDENT WHO SCORED LOWEST MARK ")
print(students[np.argmin(total)],":",np.min(total))
print("-" * 50)
print("SUBJECT WITH HIGHEST AVERAGE")
print(subjects[np.argmax(avg_sub)],":",np.max(avg_sub))
print("-" * 50)
print("PASSED STUDENTS")
for i in range(len(students)):
    if avg[i]>=60:
        print(students[i],": PASSED")
print("-" * 50)
print("FAILED STUDENTS")       
for i in range(len(students)):
    if avg[i]<60:
        print(students[i],": FAILED") 
print("-" * 50)
print("RANK OF STUDENTS")  
sorted_in=np.argsort(total)
for i in range(len(students)):
    print(students[sorted_in])
    
      
```
