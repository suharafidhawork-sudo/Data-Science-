```python
bonus_df = pd.DataFrame({
    'Emp_ID':[101,103,105,107,109],
    'Bonus':[5000,7000,4000,6000,8000]
})
```

```python
result = df.merge(bonus_df, on='Emp_ID', how='inner')
result
```

```python
r2 = df.merge(bonus_df, on='Emp_ID', how='left')
r2
```

```python
r3= df.merge(bonus_df, on='Emp_ID', how='right')
r3
```

```python
r4=df.merge(bonus_df, on='Emp_ID', how='outer')
r4
```

```python
result2 = df.merge(bonus_df, on='Emp_ID', how='inner')[['Emp_ID', 'Name', 'Salary', 'Bonus']]
print(result2)
```

```python
avg_salary_dept = df.pivot_table(values='Salary', index='Department', aggfunc='mean')
avg_salary_dept
```

```python
avg_salary_city = df.pivot_table(values='Salary', index='City', aggfunc='mean')
avg_salary_city
```

```python
pivot_city_dept = df.pivot_table(values='Salary', index='City', columns='Department', aggfunc='mean')
pivot_city_dept
```

```python
pivot_sum_sal=df.pivot_table(values='Salary',index='Department',aggfunc='sum')
pivot_sum_sal
```

```python
emp_count_city = df.pivot_table(values='Emp_ID', index='City', aggfunc='count')
emp_count_city
```

```python
above_avg = df[df['Salary'] > df['Salary'].mean()]
above_avg
```

```python
second_highest = df['Salary'].drop_duplicates().nlargest(2).iloc[-1]
print(second_highest)
```

```python
third_highest = df['Salary'].drop_duplicates().nlargest(3).iloc[-1]
print(third_highest)
```
