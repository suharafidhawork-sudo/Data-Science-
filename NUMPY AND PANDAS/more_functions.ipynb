```python
import pandas as pd
```

```python
# creating a dataframe using multiple data frames

d1=pd.DataFrame({'name':['fidha','anu','paru','krie'],
                 'age':[22,22,23,23]})
d2=pd.DataFrame({'name':['anurag','karthik'],
                 'age':[23,23]})
d4=pd.DataFrame({'city':['ekm','ekm','tcr','tcr','mal','path']})
d3=pd.concat([d1,d2],ignore_index=True)
d5=pd.concat([d3,d4],axis=1)
d5['department']=['ai','ds','ai','ds','ai','ai']
d5['salary']=[23000,450000,33400,50000,45000,60000]
d5
```

```python
#UNIQUE VALUES IN A DATA SET
d5['city'].unique()#returns the unique labels in that specific column
d5['city'].nunique()#returns the number of unique labels 
```

```python
# FUNCTION TO COUNT THE LABELS 
print(d5['city'].value_counts())#it counts the the label
d5['city'].value_counts(normalize=True)#---> shows the propotions/perctange 
```

```python
#GROUP BY FUNCTIONS
#used to group data to perform aggreate functions
#eg: calculating total salary of each department
print(d5.groupby('department')['salary'].sum())
d5.groupby('department')['salary'].agg(['mean','sum','count'])
```

```python
d5.groupby(['department','city'])['salary'].sum()
```

```python
#pivot --->used to apply the aggrate function and present it in a tabular form
d5['gender']=['f','f','f','f','m','m']
pivot=pd.pivot_table(d5,index=['department'],columns='gender',values='salary',aggfunc='mean')
print(pivot)
```
