```python
import pandas as pd
```

```python
s=pd.Series([10,20,30,40,50])
print(s)#---->labeled array
```

```python
#creating series with custom index
data=[10,20,30,40]
s1=pd.Series(data,index=['a','b','c','d'])
print(s1)
```

```python
#creating series from dictioary
d={'fidha':78,'anu':89,'krishnendu':34,'parvathy':44}
s2=pd.Series(d)
print(s2)
```

```python
#creating series using scalar values
s3=pd.Series(10,index=['a','b','c','d'])
print(s3)

```

```python
#accessing elements from series
print(s2)
#by label
print(s2['fidha'])
#by index
print(s3[0:1])
#by slicing
print(s3['b':'c'])
```
