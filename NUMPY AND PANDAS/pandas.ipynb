```python
import pandas as pd
```

```python
s=pd.Series([10,20,30,40,50])
print(s)#---->labeled array
```

```python
#creating series with custom index
data=[10,20,30,40]
s1=pd.Series(data,index=['a','b','c','d'])
print(s1)
```

```python
#creating series from dictioary
d={'fidha':78,'anu':89,'krishnendu':34,'parvathy':44}
s2=pd.Series(d)
print(s2)
```

```python
#creating series using scalar values
s3=pd.Series(10,index=['a','b','c','d'])
print(s3)

```

```python
#accessing elements from series
print(s2)
#by label
print(s2['fidha'])
#by index
print(s3[0:1])
#by slicing
print(s3['b':'c'])
```

```python
#indxmax() and idxmin()-->returns the label of max and min value in that series
s4=pd.Series([1,2,3,4],index=['a','b','c','d'])
print(s4.idxmax())
print(s4.idxmin())
```

```python
#element wise operations
s5=pd.Series([2,3,4,5])
print("SERIES \n",s5)
print("adding 10 to each element\n",s5+10)
print("multiplying with 2\n",s5*2)
print("square of elements\n",s5**2)
print("divide by 2:\n",s5/2)
#and so on you can use any arthemetic operator like these
```

```python
#ELEMENT WISE OPERATION BETWEEN TWO SERIES
a=pd.Series([10,20,30],index=['a','b','c'])
b=pd.Series([1,2,3],index=['a','b','c'])#--->should be of same index
print("PERFORMING ARTHEMETIC OPERATION BETWEEN SERIES")
print("ADD A AND B :\n",a+b)
print("SUB A AND B:\n",a-b)
print("DIV A AND B:\n",a/b)
print("MULTI A AND V:\n",a*b)
```

```python
#ELEMENT WISE COMPARISON OPERATOR
c=pd.Series([20,40,50,30])
print(c>20)#--->returns boolean value
print(c==40)
print(c%2==0)
```

```python
#OTHER FUNCTIONS ON SERIES
d=pd.Series([10,20,30,40])
print("mean:",d.mean())
print("median:",d.median())
print("standar deviation:",d.std())
print("sum:",d.sum())
print("mode:",d.mode())#--->most frequent value
print("min:",d.min())
print("varience:",d.var())
print("STATISTICAL SUMMARY:\n",d.describe())#---->to get the statistical summary of data
```
