```python

#CREATING A DATA FRAME

#from dictionary
import pandas as pd
data={'name':['fidha','anu'],
      'age':[22,23],
      'city':['ekm','ekm']}
df=pd.DataFrame(data)
print(df)
```

```python
#from list of list
data=[['fidha',22,67],['anu',23,56],['krie',23,65]]
df=pd.DataFrame(data,columns=['name','age','mark'])
print(df)
```

```python
#from dictionary of series
data={'marks':pd.Series([20,45,89]),
      'grade':pd.Series([5,6,7])}
df=pd.DataFrame(data)
print(df)
df.index=['a','b','c']#---->change the index value
print(df)
df["marks"]
df.columns#--> shows the column name
df.ndim#---diemension
df.shape#--> no of rows and col
df.size#--->number of elements in the data
```

```python
#OPERATORS
print("\n original data frame:\n")
print(df)
print("operation")
print("\nadd 10 to a dataframe:\n",df+10 )
data2={"marks":[10,20,30],
        "age":[22,45,34]}
df2=pd.DataFrame(data2)
print("\n between to dataframe:\n")
print(df+df2)
```

```python
df1=pd.DataFrame({'A':[20,30,40],'B':[1,2,4]})
print("dataset 1")
print(df1)
df2=pd.DataFrame({'A':[1,2,3],'B':[34,78,23]})
print("dataset 2")
print(df2)
```

```python
#LOGICAL OPERATORS

data={'name':["fidha","anu","krie","paru"],
      'age':[22,23,23,22],
      'mark':[92,84,93,88]}
df3=pd.DataFrame(data)
df3
```

```python
#boolean expressions
df3[(df3['age']> 18)|(df3['mark']>90)]#----> AND
```

```python
df3[(df3['age']> 18)&(df3['mark']>90)]#----> OR
```

```python
df3[~(df3['age']< 18)]#--->NOT 
```

```python
df3.sum()#---> sum of of each column
df3.sum(axis=1,numeric_only=True)#--->sum of row/takes only the numerical data in the dataset
df3.describe(include='all')#---->statistical summary of data
```

```python
df3.mean(numeric_only=True)#-->mean of each #col
df3.mean(numeric_only=True,axis=1)#---mean of each row
df3["mark"].mean()#---gives mean of that particular col, axis=1 then gves the row
df3.std(numeric_only=True)#--colwise std/axis=1 then rowise
df3.min(numeric_only=True)#min elements in the columns in the data set
df3.var(numeric_only=True)#varience measure of each column
```
