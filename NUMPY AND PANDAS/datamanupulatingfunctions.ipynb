```python
df.drop('Score',axis=1)#drops the column score and creates a new data frame if inplace=true then it changes the existing dataset
```

```python
#filtering the data
df[df['Age']<40]#using boolean filtering/one condition
df[(df['Age']==40)&(df['Score']>4.2)]#using logica operators/multiple conditions
```
