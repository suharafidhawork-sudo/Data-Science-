```python
import pandas as pd
```

```python
d1=pd.DataFrame({'name':['fidha','anu','paru','krie'],
                 'age':[22,22,23,23]})
d2=pd.DataFrame({'name':['anurag','karthik'],
                 'age':[23,23]})
d4=pd.DataFrame({'city':['ekm','ekm','tcr','tcr','mal','path']})
```

```python
#VERTICAL BINDING
#stacking a dataframe on top of another

d3=pd.concat([d1,d2],axis=0,ignore_index=True)
d3
```

```python
#HORIZONATAL BINDIN
d5=pd.concat([d3,d4],axis=1)
print(d5)
```

```python
#MERGED
#to join dataframe based on a condition or like based a column
#similar to sql inner,left,outer joins

print(d1)
d1['ID']=[1,2,3,4]
d1
#based on this id joing another data frame
d6=pd.DataFrame({'ID':[1,2,4],
                 'city':['tcr','ekm','ekm']})

#inner join
df_inner=pd.merge(d1,d6,how='inner',on='ID')
df_inner#joins two data frame where ot has matching ids
df_outer=pd.merge(d1,d6,how='outer',on='ID')
df_outer#joings all the dataset 
```
